# Violin plots: whole-body model vs thigh model

Compares segmentation metrics from `MuscleMap_results_WB` (whole-body model) against `MuscleMap_results_thigh` (thigh model) for each muscle group.

```python
import os
from dissector.evaluation import violin_compare

wb_root    = "../MuscleMap_results_WB"
thigh_root = "../MuscleMap_results_thigh"
```

## Right gracilis

```python
violin_compare(
    csv_list_a=[os.path.join(wb_root, "df_r_gracilis.csv")],
    csv_list_b=[os.path.join(thigh_root, "df_r_gracilis_thigh.csv")],
    label_a="Whole-body",
    label_b="Thigh",
    save_path="violin_r_gracilis.png",
)
```

## Left gracilis

```python
violin_compare(
    csv_list_a=[os.path.join(wb_root, "df_l_gracilis.csv")],
    csv_list_b=[os.path.join(thigh_root, "df_l_gracilis_thigh.csv")],
    label_a="Whole-body",
    label_b="Thigh",
    save_path="violin_l_gracilis.png",
)
```

## Right sartorius
