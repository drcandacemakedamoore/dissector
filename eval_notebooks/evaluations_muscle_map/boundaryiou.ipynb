```python
def scroll_side(mask: np.ndarray):
    assert mask.ndim == 3, "Expected 3D array"

    h, w, d = mask.shape  # (652, 652, 65)

    def show_slice(i):
        plt.figure(figsize=(5, 5))

        # Sagittal slice
        img = mask[i, :, :]

        # Fix orientation so "up" is superior
        img = np.rot90(img)

        plt.imshow(img, cmap="gray", vmin=0, vmax=1)
        plt.title(f"Sagittal slice {i}/{h-1}")
        plt.axis("off")
        plt.show()

    widgets.interact(
        show_slice,
        i=widgets.IntSlider(min=0, max=h-1, step=1, value=h//2, description="Slice")
    )
```

```python
scroll_side(example_bound_b)
```
