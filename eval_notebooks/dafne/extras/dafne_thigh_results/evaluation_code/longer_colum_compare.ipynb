# MuscleMap Thigh — Extended Metrics (_LONGER)

Reads the existing 6-metric CSVs from `MuscleMap_results_thigh/`, loads each
MuscleMap thigh NIfTI prediction, computes the four missing metrics
(`binary_cross_entropy`, `boundary_iou_3d`, `inter_slice_dice_pred`,
`inter_slice_dice_gt`), and saves `*_LONGER.csv` files in the same folder.

Run this notebook from `evaluation_code/` so relative paths resolve correctly.

```python
# libraries
import os
import re
import numpy as np
import pandas as pd
import SimpleITK as sitk
from dissector.evaluation import binary_cross_entropy, boundary_iou_3d, inter_slice_dice

BOUNDARY_DISTANCE = 1

GT_BASE   = os.path.join("..", "..", "myosegmenTUM")
PRED_DIR  = os.path.join("..", "..", "MuscleMap_thigh_segs")
CSV_IN    = os.path.join("..", "..", "MuscleMap_results_thigh")
CSV_OUT   = CSV_IN   # _LONGER.csv files go in the same folder
```

## Left Gracilis  (GT label 1 · pred label 7151)

```python
existing = pd.read_csv(os.path.join(CSV_IN, "df_l_gracilis_thigh.csv"), index_col=0)
```

```python

extra_rows = []
for _, row in existing.iterrows():
    gt_path   = os.path.join(GT_BASE, row["image"])
    pred_path = os.path.join(PRED_DIR, row["pred_label"])

    gt_image      = sitk.ReadImage(gt_path)
    segment_image = sitk.ReadImage(pred_path)
    segment_image.CopyInformation(gt_image)

    gt   = sitk.Cast(gt_image      == 1,        sitk.sitkUInt8)
    pred = sitk.Cast(segment_image == 7151,  sitk.sitkUInt8)

    gt_arr   = sitk.GetArrayFromImage(gt).astype(float)
    pred_arr = sitk.GetArrayFromImage(pred).astype(float)

    bce      = binary_cross_entropy(gt_arr, pred_arr)
    biou     = boundary_iou_3d(BOUNDARY_DISTANCE, gt_arr, pred_arr)
    isd_pred = inter_slice_dice(pred_arr)
    isd_gt   = inter_slice_dice(gt_arr)

    extra_rows.append({
        "L_gracilis_binary_cross_entropy":  bce,
        "L_gracilis_boundary_iou_3d":       biou,
        "L_gracilis_inter_slice_dice_pred": isd_pred,
        "L_gracilis_inter_slice_dice_gt":   isd_gt,
    })

extra_df = pd.DataFrame(extra_rows, index=existing.index)
longer   = pd.concat([existing, extra_df], axis=1)
longer.to_csv(os.path.join(CSV_OUT, "df_l_gracilis_thigh_LONGER.csv"))
longer
```
